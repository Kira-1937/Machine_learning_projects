```python
df.drop(['employment_occupation','employment_industry'],axis=1,inplace=True)
df.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(handle_unknown='ignore',sparse_output=False)
```

```python
cols=['education','income_poverty' ,'marital_status' ,'rent_or_own' , 'employment_status']
encode=ohe.fit_transform(df[cols])
encoded_df=pd.DataFrame(encode,columns=ohe.get_feature_names_out(cols),index=df.index)
df=pd.concat([df.drop(cols,axis=1),encoded_df],axis=1)
```

```python
from sklearn.preprocessing import OneHotEncoder 
encoder =OneHotEncoder(sparse_output=False,handle_unknown='ignore')
col=['age_group','race','sex','hhs_geo_region','census_msa']
encode=encoder.fit_transform(df[col])
encoded_df=pd.DataFrame(encode,columns=encoder.get_feature_names_out(col),index=df.index)
df=pd.concat([df.drop(col,axis=1),encoded_df],axis=1)
df.info()
```

```python
df_test=df[df['h1n1_vaccine'].isna()]
df_train=df[~df['h1n1_vaccine'].isna()]
```

```python
df_train.shape
```

```python
df_test.shape
```

```python
from sklearn.model_selection import train_test_split
x_train=df_train.drop(['h1n1_vaccine','seasonal_vaccine'],axis=1)
y1_train=df_train['h1n1_vaccine']
y2_train=df_train['seasonal_vaccine']
```

```python
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
scaler.fit_transform(x_train)
```

```python
# feature selection 
from sklearn.feature_selection import mutual_info_classif
mi_y1=mutual_info_classif(x_train,y1_train)
mi_y2=mutual_info_classif(x_train,y2_train)
print(mi_y1,mi_y2)

```

```python
mi_y1_series=pd.Series(mi_y1,index=x_train.columns)
mi_y2_series=pd.Series(mi_y2,index=x_train.columns)
print(mi_y1_series.sort_values(ascending=False))
print(mi_y2_series.sort_values(ascending=False))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_mi_scores(scores):
    scores = scores.sort_values(ascending=True)
    width = np.arange(len(scores))
    ticks = list(scores.index)
    plt.barh(width, scores)
    plt.yticks(width, ticks)
    plt.title("Mutual Information Scores")


plt.figure(dpi=100, figsize=(12, 16))
plot_mi_scores(mi_y1_series.sort_values(ascending=False))
```

```python
important_features_y1 = mi_y1_series[mi_y1_series > 0.001].index
important_features_y2 = mi_y2_series[mi_y2_series > 0.001].index
```

```python
print(important_features_y1)
print(important_features_y2)
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc1=RandomForestClassifier(n_estimators=100,n_jobs=10)
rfc2=RandomForestClassifier(n_estimators=100,n_jobs=10)
rfc1.fit(x_train[important_features_y1],y1_train)
rfc2.fit(x_train[important_features_y2],y2_train)
```

```python
y1_pred=rfc1.predict_proba(df_test[important_features_y1])
y2_pred=rfc2.predict_proba(df_test[important_features_y2])
```

```python
y1_pred
```

```python
result_df = pd.concat([
    df_test['respondent_id'],
    pd.DataFrame(y1_pred[:,1], columns=['h1n1_vaccine']),
    pd.DataFrame(y2_pred[:,1], columns=['seasonal_vaccine'])
], axis=1)

result_df.head()
```
